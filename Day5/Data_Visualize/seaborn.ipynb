```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset('penguins')
penguins.head()
```

```python
penguins["species"].value_counts()
```

```python
penguins["island"].value_counts()
```

```python
sns.set_style("darkgrid")
sns.scatterplot(data = penguins, x="flipper_length_mm", y = "body_mass_g", hue = "sex")
```

```python
sns.stripplot(data=penguins, x= "species", y="body_mass_g", hue="sex", jitter= True)
```

```python
# Swarm plot

sns.swarmplot(data=penguins, x="species", y="body_mass_g", hue="sex")
```

```python
# histogram

sns.set_context("notebook")
sns.histplot(data=penguins, x="body_mass_g", hue="sex", multiple="stack")
```

```python
# regression plot:

sns.regplot(data=penguins, x="body_mass_g", y="flipper_length_mm", color="orange")
```

```python
# line plot:
sns.lineplot(data=penguins,  x="body_mass_g", y="flipper_length_mm", hue="island", style="sex")
```

```python
# join plots

sns.jointplot(data=penguins, x="body_mass_g", y="flipper_length_mm")
```

```python
sns.jointplot(data = penguins, x="body_mass_g", y="flipper_length_mm", hue="sex")
```

```python
sns.jointplot(data=penguins,  x="body_mass_g", y="flipper_length_mm",hue="sex", kind="kde")
```

```python
sns.jointplot(data=penguins,  x="body_mass_g", y="flipper_length_mm", kind="reg")
```

```python
# bar plot

sns.barplot(data=penguins, x="species", y="body_mass_g", hue="sex", palette=["skyblue", "green"])
```

```python
# countplot

sns.countplot(data=penguins, x="species", hue="island", palette="Set2")
```

```python
# box plot

sns.boxplot(data=penguins,  x="species", y="body_mass_g",hue="sex")
```

```python
# violin plot
sns.set_style("whitegrid")
sns.violinplot(data=penguins,  x="species", y="body_mass_g",hue="sex")
```

```python
sns.violinplot(data=penguins,  x="species", y="body_mass_g",hue="sex", split=True)
```

```python
sns.violinplot(data=penguins,  x="species", y="body_mass_g")
sns.swarmplot(data=penguins,  x="species", y="body_mass_g", color="orange", size=3)
```

```python
# kde plot
sns.kdeplot(data=penguins,  x="body_mass_g", hue="species", palette="pastel", fill=True)
sns.despine()
```

```python
# Rug plot:
sns.rugplot(data=penguins, x = "body_mass_g", hue="species", palette="Set2")
```

```python
# Rug plot:
sns.rugplot(data=penguins, y = "body_mass_g", hue="species", palette="Set2", height=0.5)
```

```python
# pair plot:

sns.pairplot(data=penguins)
```

```python
sns.pairplot(data=penguins, hue="sex", palette="tab10")
```

```python
# pair grid -> subplot grid for plotting pairwise relationship in a dataset

graph = sns.PairGrid(data=penguins, hue="sex", palette="tab10")

graph.map_upper(sns.scatterplot)
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot)

graph.add_legend()
```
