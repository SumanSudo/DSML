```python
# Pie chart:

count = df["dept"].value_counts()
```

```python
count
```

```python
explode = [0] * len(count)
explode[1] = 0.1   # Explode the second slice

plt.pie(
    count,
    labels=count.index,
    autopct="%1.1f%%",
    explode=explode
)
plt.axis("equal")
plt.show()
```

```python
# Cpuntplot
plt.bar(count.index, count, color=["green", "red", "black", "yellow"])
plt.show()
```

```python
import numpy as np

df["Age"] = np.random.randint(18, 61, size=14)
```

```python
df.head()
```

```python
# Scatter Plot: 
plt.scatter(df["Age"], df["Salary"], color="orange")
plt.show()
```

```python
sort_age = df.sort_values("Age")
```

```python
# line plot:
plt.plot(sort_age["Age"], df["Salary"], color="red", marker="*", linestyle=":", linewidth="2")
plt.grid()
plt.show()
```

```python
# Bar Chart:
plt.bar(sort_age["Age"], df["Salary"], color="green")
plt.show()
```

```python
# Bivariate: Numerical - Categorical
```

```python
hr_sal = df[df["dept"] == "HR"]["Salary"]
hr_sal
```

```python
fin_sal = df[df["dept"] == "Finance"]["Salary"]
fin_sal
```

```python
it_sal = df[df["dept"] == "IT"]["Salary"]
it_sal
```

```python
ex_sal = df[df["dept"] == "Exam"]["Salary"]
ex_sal
```

```python
# Boxplot:
plt.boxplot([hr_sal, fin_sal, ex_sal, it_sal], labels=["HR", "FIN", "EX", "IT"])
plt.show()
```

```python
# pie chart

salary_by_dept = df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept, labels=salary_by_dept.index, autopct = "%1.2f", shadow=True,  explode=[0, 0.1, 0, 0])
plt.axis("equal")
plt.show()
```

```python
# bar plot

hr_mean = sum(hr_sal)/len(hr_sal)
fin_mean = sum(fin_sal)/len(fin_sal)
ex_mean = sum(ex_sal)/len(ex_sal)
it_mean = sum(it_sal)/len(it_sal)
```

```python
plt.barh(
    ["HR", "IT", "Finance", "Exam"],
    [hr_mean, it_mean, fin_mean, ex_mean],
    color=["green", "red", "blue", "yellow"]
)
plt.show()
```

```python
df["experiance"] = np.random.randint(1, 6, size=14)
```

```python
# Bubble Plot:

plt.scatter(df["Age"], df["Salary"], s=df["experiance"]*20, color="orange", edgecolors="black")
plt.title("Age vs Sal vs Ex")
plt.show()
```

```python
# 2 numerical and 1 categorical column
```

```python
dept_codes = df["dept"].astype("category").cat.codes

plt.scatter(
    df["Age"],
    df["Salary"],
    c=dept_codes,
    cmap="rainbow"
)

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary vs Dept")

plt.colorbar(label="Department")
plt.show()
```

```python
color = { "HR" : "yellow", "IT": "blue", "Finance":"green", "Exam":"black"}

for dept, color in color.items():
    df_dept = df[df["dept"] == dept]
    plt.scatter(df_dept["Age"], df_dept["Salary"], c=color, label=dept)

    plt.legend()
```

```python
# Object Oriented API
```

```python
fig, axs = plt.subplots(1, 3, figsize=(11, 5))

# Line plot
axs[0].plot(
    sort_age["Age"],
    sort_age["Salary"],
    color="red",
    marker="*",
    linewidth=2,
    markersize=5
)
axs[0].grid()
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")


# Bar plot
axs[1].bar(
    ["HR", "IT", "Finance", "Exam"],
    [hr_mean, it_mean, fin_mean, ex_mean],
    color=["green", "red", "blue", "yellow"]
)
axs[1].set_title("Bar Plot")
axs[1].set_xlabel("Department")
axs[1].set_ylabel("Average Salary")


# Scatter plot
axs[2].scatter(
    df["Age"],
    df["Salary"],
    color="purple"
)
axs[2].set_title("Scatter Plot")
axs[2].set_xlabel("Age")
axs[2].set_ylabel("Salary")


plt.tight_layout()

plt.savefig("Obj_ori_api_test.png")
plt.show()
```

```python
fig = plt.figure(figsize=(11, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    df["Age"],
    df["Salary"],
    df["experiance"],
    color="orange"
)

ax.set_xlabel("Age")
ax.set_ylabel("Salary")
ax.set_zlabel("Experience")

ax.set_title("3D Plot: Age vs Salary vs Experience")

plt.show()
```
